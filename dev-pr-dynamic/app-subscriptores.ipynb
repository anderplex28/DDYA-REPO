## 6) Código funcional **[15 pts]**
Incluye tu implementación completa.

No pegues código incompleto; debe ser ejecutable.


```python
subs = [0, 1]


def memor(n):
    if n < len(subs):
        return subs[n]

    for i in range(len(subs), n + 1):
        subs.append(subs[i - 1] + subs[i - 2])

    return subs[n]


def main():
    n = int(input("Ingrese el número de mes al cual desea conocer la cantidad de suscriptores: "))

    if n >= 0:
        subscr = memor(n)
        print("La cantidad de suscriptores es:", subscr)
    else:
        des = input("El valor ingresado no es válido. ¿Desea reiniciar el proceso? ")

        if des.lower() == "si":
            main()


main()
```
