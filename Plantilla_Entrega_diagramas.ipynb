```python
#Merge sort Julian Sanchez y Anderson Varela
def merge_sort(lista):
      if len(lista) <= 1:
            return lista
      
      mitad = len(lista) // 2
      iz = merge_sort(lista[:mitad])
      de = merge_sort(lista[mitad:])
      return merge(iz,de)
def merge(izquierda, derecha):
    resultado = []

    i = 0
    j = 0

    while i < len(izquierda) and j < len(derecha):
        if izquierda[i] < derecha[j]:
            resultado.append(izquierda[i])
            i += 1
        else:
            resultado.append(derecha[j])
            j += 1

    resultado.extend(izquierda[i:])
    resultado.extend(derecha[j:])
    return resultado

def productos(n,lista):
      vc = 0
      while vc < n:
            producto = int(input(f"Ingrese codigo del producto{vc + 1}"))
            lista.append(producto)
            vc+=1
      return lista
def busqueda_binaria(lista, objetivo):
    izquierda = 0
    derecha = len(lista) - 1

    while izquierda <= derecha:
        medio = (izquierda + derecha) // 2

        if lista[medio] == objetivo:
            return True

        elif lista[medio] < objetivo:
            izquierda = medio + 1

        else:
            derecha = medio - 1

    return False
def main():
      n = int(input("Ingrese la cantidad de productos: "))
      lista = []

      productos(n, lista)

      lista_ordenada = merge_sort(lista)

      print("Ordenado:", lista_ordenada)

      codigo = int(input("Ingrese el código que desea buscar: "))

      encontrado = busqueda_binaria(lista_ordenada, codigo)

      if encontrado:
          print(f"El producto con código {codigo} fue encontrado.")
      else:
          print(f"El producto con código {codigo} no se encuentra registrado.")


main()

```
