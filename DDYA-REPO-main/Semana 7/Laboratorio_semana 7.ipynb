### 5) Análisis de complejidad **[15 pts]**
**Justificacion**

Los algoritmos analizados son insertar_bst e insertar_avl. Ambos son funciones recursivas que descienden por el árbol comparando el valor a insertar contra el valor del nodo actual, y avanzan hacia la izquierda o hacia la derecha según corresponda, hasta encontrar una posición vacía (None) donde crear el nuevo nodo.

**Funcion T(n)**

Sea h la altura del árbol en el momento de la inserción. La recursión de cada inserción tiene la forma:

resolviendo por sustitucion :

T(h) = c·h + c  =  O(h)


BST: h puede variar entre log₂(n) (árbol balanceado) y n (árbol degenerado) → T(n) = O(h), con h ∈ [log₂n, n].
AVL: la propiedad de balance estricto (|factor de balance| ≤ 1 en cada nodo) garantiza matemáticamente que h ≤ 1.44·log₂(n) siempre → T(n) = O(log n) en todos los casos.

**Mejor y peor caso**

Mejor caso — O(log n)

Justificación: Ocurre cuando los códigos se insertan en un orden que mantiene el árbol equilibrado (por ejemplo, orden aleatorio o ya balanceado, como en la prueba oficial [30, 20, 10, 25, 28, 27, 5, 50]). En este escenario, cada comparación descarta aproximadamente la mitad de los nodos restantes, por lo que la altura del árbol crece proporcional a log₂(n). Tanto BST como AVL alcanzan esta cota en el mejor caso, porque si el árbol ya está balanceado, el AVL no necesita rotar.

image.png

Peor caso — O(n) para BST, O(log n) para AVL

Justificación: Ocurre cuando los códigos llegan ya ordenados (ascendente o descendente), como en la prueba propia del notebook con [10, 20, 30, 40, 50]. En el BST, cada nuevo valor es siempre mayor (o menor) que todos los anteriores, así que se inserta siempre en el mismo extremo, formando una cadena lineal equivalente a una lista enlazada: la altura pasa a ser n, y por tanto la inserción cuesta O(n). El AVL, en cambio, detecta el desbalance en cada inserción (factor de balance fuera de {-1, 0, 1}) y aplica una rotación en O(1) para corregirlo, manteniendo la altura acotada por 1.44·log₂(n) sin importar el orden de entrada — por eso su curva sigue siendo logarítmica incluso en este escenario adverso.

```python
import matplotlib.pyplot as plt

def crear_nodo(valor):
    return {
        "valor": valor,
        "izquierda": None,
        "derecha": None,
        "altura": 1
    }



# ÁRBOL BINARIO DE BÚSQUEDA (BST)


def insertar_bst(raiz, valor):
    if raiz is None:
        return crear_nodo(valor)

    if valor < raiz["valor"]:
        raiz["izquierda"] = insertar_bst(raiz["izquierda"], valor)
    elif valor > raiz["valor"]:
        raiz["derecha"] = insertar_bst(raiz["derecha"], valor)

    return raiz



# FUNCIONES PARA EL ÁRBOL AVL


def obtener_altura(nodo):
    if nodo is None:
        return 0
    return nodo["altura"]


def actualizar_altura(nodo):
    nodo["altura"] = 1 + max(
        obtener_altura(nodo["izquierda"]),
        obtener_altura(nodo["derecha"])
    )


def obtener_factor_balance(nodo):
    if nodo is None:
        return 0
    return obtener_altura(nodo["izquierda"]) - obtener_altura(nodo["derecha"])


def rotacion_derecha(y):
    x = y["izquierda"]
    temporal = x["derecha"]

    x["derecha"] = y
    y["izquierda"] = temporal

    actualizar_altura(y)
    actualizar_altura(x)

    return x


def rotacion_izquierda(x):
    y = x["derecha"]
    temporal = y["izquierda"]

    y["izquierda"] = x
    x["derecha"] = temporal

    actualizar_altura(x)
    actualizar_altura(y)

    return y


def insertar_avl(raiz, valor):
    if raiz is None:
        return crear_nodo(valor)

    if valor < raiz["valor"]:
        raiz["izquierda"] = insertar_avl(raiz["izquierda"], valor)
    elif valor > raiz["valor"]:
        raiz["derecha"] = insertar_avl(raiz["derecha"], valor)
    else:
        return raiz

    actualizar_altura(raiz)
    balance = obtener_factor_balance(raiz)

    # Caso LL
    if balance > 1 and valor < raiz["izquierda"]["valor"]:
        return rotacion_derecha(raiz)

    # Caso RR
    if balance < -1 and valor > raiz["derecha"]["valor"]:
        return rotacion_izquierda(raiz)

    # Caso LR
    if balance > 1 and valor > raiz["izquierda"]["valor"]:
        raiz["izquierda"] = rotacion_izquierda(raiz["izquierda"])
        return rotacion_derecha(raiz)

    # Caso RL
    if balance < -1 and valor < raiz["derecha"]["valor"]:
        raiz["derecha"] = rotacion_derecha(raiz["derecha"])
        return rotacion_izquierda(raiz)

    return raiz




def preorden(raiz, resultado):
    if raiz is not None:
        resultado.append(raiz["valor"])
        preorden(raiz["izquierda"], resultado)
        preorden(raiz["derecha"], resultado)


def inorden(raiz, resultado):
    if raiz is not None:
        inorden(raiz["izquierda"], resultado)
        resultado.append(raiz["valor"])
        inorden(raiz["derecha"], resultado)


def postorden(raiz, resultado):
    if raiz is not None:
        postorden(raiz["izquierda"], resultado)
        postorden(raiz["derecha"], resultado)
        resultado.append(raiz["valor"])



# VISUALIZAR ÁRBOL


def dibujar_arbol(raiz, titulo):
    posiciones = {}
    conexiones = []

    def calcular_posiciones(nodo, x, y, separacion):
        if nodo is None:
            return
        posiciones[id(nodo)] = (x, y)

        if nodo["izquierda"] is not None:
            nuevo_x, nuevo_y = x - separacion, y - 1
            conexiones.append(((x, y), (nuevo_x, nuevo_y)))
            calcular_posiciones(nodo["izquierda"], nuevo_x, nuevo_y, separacion / 2)

        if nodo["derecha"] is not None:
            nuevo_x, nuevo_y = x + separacion, y - 1
            conexiones.append(((x, y), (nuevo_x, nuevo_y)))
            calcular_posiciones(nodo["derecha"], nuevo_x, nuevo_y, separacion / 2)

    calcular_posiciones(raiz, 0, 0, 4)

    plt.figure(figsize=(10, 6))
    plt.title(titulo, fontsize=14, fontweight="bold")

    for inicio, fin in conexiones:
        plt.plot([inicio[0], fin[0]], [inicio[1], fin[1]], color="black", linewidth=2)

    def dibujar_nodos(nodo):
        if nodo is None:
            return
        x, y = posiciones[id(nodo)]
        plt.scatter(x, y, s=1200, color="lightblue", edgecolors="black", linewidth=2, zorder=2)
        plt.text(x, y, str(nodo["valor"]), ha="center", va="center", fontsize=11, fontweight="bold", zorder=3)
        dibujar_nodos(nodo["izquierda"])
        dibujar_nodos(nodo["derecha"])

    dibujar_nodos(raiz)
    plt.axis("off")
    plt.show()
```

```python
from random import randint
import time
import matplotlib.pyplot as plt

def bench_arboles(n: int, step: int = 100) -> tuple[list, list]:
    print("Generando datos de rendimiento para BST y AVL...", end='')
    resultados_bst = []
    resultados_avl = []

    for size in range(50, n + 1, step):
        datos = [randint(1, 100000) for _ in range(size)]

        # --- 1. Medir tiempo de inserción en BST ---
        ti = time.perf_counter()
        bst = None
        for val in datos:
            bst = insertar_bst(bst, val)
        tf = time.perf_counter()
        delta_bst = (tf - ti) * 1000
        resultados_bst.append((size, delta_bst))

        ti = time.perf_counter()
        avl = None
        for val in datos:
            avl = insertar_avl(avl, val)
        tf = time.perf_counter()
        delta_avl = (tf - ti) * 1000
        resultados_avl.append((size, delta_avl))

        print(".", end="")

    print(" ¡Completado!")
    return resultados_bst, resultados_avl


def dibujar_comparativa(res_bst, res_avl, dimensiones=(10, 6)):
    x_bst = [size for size, timing in res_bst]
    y_bst = [timing for size, timing in res_bst]

    x_avl = [size for size, timing in res_avl]
    y_avl = [timing for size, timing in res_avl]

    assert len(x_bst) == len(y_bst) and len(x_avl) == len(y_avl)

    plt.figure(figsize=dimensiones)
    plt.plot(x_bst, y_bst, label="Árbol BST", color="crimson", linewidth=2, marker='o')
    plt.plot(x_avl, y_avl, label="Árbol AVL", color="dodgerblue", linewidth=2, marker='s')

    plt.xlabel("Tamaño de la entrada (Cantidad de códigos $n$)", fontsize=12)
    plt.ylabel("Tiempo de ejecución en milisegundos", fontsize=12)
    plt.title("Prueba de Desempeño: Tiempo de Inserción BST vs AVL", fontsize=14, fontweight="bold")
    plt.legend(fontsize=11)
    plt.grid(True, linestyle="--", alpha=0.6)
    plt.show()



# EJECUCIÓN DE LA PRUEBA DE DESEMPEÑO

tiempos_bst, tiempos_avl = bench_arboles(1500, step=100)


dibujar_comparativa(tiempos_bst, tiempos_avl)
```
